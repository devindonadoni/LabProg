```python
#ESERCIZIO 1

#1.1
numeriPrimi = []
for i in range(2,100):
    primo = True
    for n in range(2, i):
        if i%n == 0:
            primo = False
    if primo: numeriPrimi.append(i)
print(numeriPrimi)

numeriPrimiComp = [
    i for i in range(2, 100)
    if all(i % n != 0 for n in range(2, i))
]
print(numeriPrimiComp)
```

```python
#1.2
potenze = [i**2 for i in range(1,10)]
print(potenze)
```

```python
#1.3
sommaComulativa = []
for i in range(1, 11):
    somma = 0
    for n in range(1, i):
        somma = somma + n
    sommaComulativa.append(somma)

print(sommaComulativa)

sommaComulativaComp = [sum(n for n in range(1, i)) for i in range(1, 11)]

print(sommaComulativaComp)
```

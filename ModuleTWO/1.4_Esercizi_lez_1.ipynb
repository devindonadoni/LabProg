### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
#1.1
numeriPrimi = []
for i in range(2,100):
    primo = True
    for n in range(2, i):
        if i%n == 0:
            primo = False
    if primo: numeriPrimi.append(i)
print(numeriPrimi)

numeriPrimiComp = [
    i for i in range(2, 100)
    if all(i % n != 0 for n in range(2, i))
]
print(numeriPrimiComp)

#1.2
potenze = [i**2 for i in range(1,10)]
print(potenze)


#1.3
sommaComulativa = []
for i in range(1, 11):
    somma = 0
    for n in range(1, i):
        somma = somma + n
    sommaComulativa.append(somma)

print(sommaComulativa)

sommaComulativaComp = [sum(n for n in range(1, i)) for i in range(1, 11)]

print(sommaComulativaComp)
```

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
import numpy as np

a = np.array([n for n in range(1, 11)])
b = np.array([i for i in range(2,11)
     if all(i%n != 0 for n in range(2, i))])
print(a.size)
print(len(a))
print(a.dtype)
print(b)
```
